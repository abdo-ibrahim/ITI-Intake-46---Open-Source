7-  Write a Python script to:
- Take a string input from the user.
- Convert the string to uppercase.
- Count the number of vowels in the string.
- Reverse the string.
- Check if the string is a **palindrome**
--  example: madam

```python
text = input("Enter a string: ")

upper_text = text.upper()
print("uppercase:", upper_text)

vowels = "aeiouAEIOU"
count = 0
for ch in text:
  if ch in vowels:
    count += 1

print("Number of vowels:", count)

reversed_text = text[::-1]
print("Reversed:", reversed_text)

if text == reversed_text:
  print("Palindrome")
else:
  print("Not Palindrome")
```

# Sets in python

### 1- Create a set of the first five even numbers. Write a script to:

- Add a new number to the set.
- Remove a number from the set.
- Check if a specific number exists in the set.
- Perform a union and intersection with another set `{2, 4, 6, 8, 10}`.


```python
even_numbers = {2, 4, 6, 8, 10 , 10}

even_numbers.add(12)

if 4 in even_numbers:
  even_numbers.remove(4)

if 6 in even_numbers:
  print("6 exists in the set")

set2 = {2, 4, 6, 8, 10}

print("union:", even_numbers.union(set2))

print("intersection:", even_numbers.intersection(set2))
```

### 2- Unique Visitors Tracker
Create a set to track unique website visitors. Write a script to:

- Add new visitors to the set.
- Check if a visitor has already visited.
- Remove a visitor from the set.


```python
visitors = set()

visitors.add("abdo")
visitors.add("ahmed")

name = input("Enter visitor name: ")

if name in visitors:
  print("visitor already visited")
else:
  print("new visitor")
  visitors.add(name)

remove_name = input("remove visitor: ")
if remove_name in visitors:
	visitors.remove(remove_name)
	print("visitor removed")

print("visitors:", visitors)
```

### 3- Set-Based Password Security Checker

## Task:
Implement a function that checks if a password is weak by comparing it to a predefined set of common passwords.

**Common passwords:**
- `"123456"`, `"password"`, `"123456789"`, `"12345"`, `"12345678"`, `"qwerty"`
- `"1234567"`, `"111111"`, `"123123"`, `"abc123"`, `"password1"`, `"1234"`


```python
def check_password(password):
  common_passwords = {
    "123456", "password", "123456789", "12345", "12345678", "qwerty",
    "1234567", "111111", "123123", "abc123", "password1", "1234"
  }

  if password in common_passwords:
    print("Weak password")
  else:
    print("Strong password")


pwd = input("Enter password: ")
check_password(pwd)
```

# Tuples in Python

## **Task 1: Working with Tuples**
Create a tuple of five different fruits. Write a script to:

- Print the first and last fruit.
- Slice the tuple to get the first three fruits.
- Count the occurrences of a specific fruit.
- Try modifying the tuple and handle the error gracefully.




```python
fruits = ("apple", "banana", "orange", "mango", "grape", "apple" , "apple")

print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])

print("First three:", fruits[:3])

fruit_name = input("Enter fruit to count: ")
print("Count:", fruits.count(fruit_name))

try:
  fruits[0] = "pear"
except TypeError:
  print("Error: Tuples cannot be modified")
```


## **Task 2: Tuple-Based Stock Price Analysis**
Given a tuple of stock prices for a week, write a script to:

- Find the highest price.
- Find the lowest price.
- Calculate the average price.
    -  `Output`:
    -	Highest Price: 162
    -	Lowest Price: 145
    -	Average Price: 154.00

```python
prices = (150, 162, 158, 145, 155, 160, 148)

highest = max(prices)
lowest = min(prices)
average = sum(prices) / len(prices)

print("Highest Price:", highest)
print("Lowest Price:", lowest)
print("Average Price:", f"{average:.2f}")
```
