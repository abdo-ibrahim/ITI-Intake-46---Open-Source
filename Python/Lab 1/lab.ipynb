3) The volume of a sphere with radius r is 4/3πr3. What is the volume of a sphere
with radius 5? Start with a variable named radius and then assign the result to
a variable named volume. Display the result. Add comments to indicate
that radius is in centimeters and volume in cubic centimeters.

```python
radius = 5
volume = (4/3) * 3.14159 * (radius ** 3)
print(f"The volume of a sphere with radius {radius} cm is {volume:.2f} cubic cm")
```

10) Prime Number Checker
Ask the user for a number and check if it's a prime number.

```python
num = int(input("Enter a number: "))

if num > 1:

  is_prime = True

  for i in range(2, int(num ** 0.5) + 1):
    if num % i == 0:
      is_prime = False
      break

  if is_prime:
    print(f"{num} is a prime number")
  else:
    print(f"{num} is not a prime number")

else:
  print(f"{num} is not a prime number")
```

11- Write a program that asks the user for:

 - The total number of students enrolled in a course

 - The desired number of students per group

The program should then calculate and display the total number of groups formed. If the students cannot be divided evenly, one group may contain fewer members than specified.

**Example output (1):**

- How many students on the course? 8
- Desired group size? 4
- Number of groups formed: 2


**Example output (2):**

- How many students on the course? 11
- Desired group size? 3
- Number of groups formed: 4


> Hint: The integer division operator // may be helpful in solving this problem.

```python
students = int(input("How many students on the course? "))
group_size = int(input("Desired group size? "))
groups = students // group_size
if students % group_size != 0:
    groups += 1
print(f"Number of groups formed: {groups}")
```

12-Write a program that prints the number of times the string 'iti'
occurs in anystring.

```python
# str = input("Enter any string: ")
# count = str.lower().count('iti')
# print(f"'iti' occurs {count} times in the string")

str = input("Enter any string: ")
count = 0
s = str.lower()
for i in range(len(s) - 2):
  if s[i:i+3] == 'iti':
    count += 1
print(f"'iti' occurs {count} times in the string")
```

13- Write a program that remove all vowels from the input word and
generate a brief version of it

---

```python
word = input("Enter a word: ")
vowels = "aeiouAEIOU"
brief = ""
for char in word:
  if char not in vowels:
    brief += char
print(f"Brief version: {brief}")
```

14- Write a program that counts up the number of vowels **[a, e, i, o,u]** contained in the string.


```python
string = input("Enter a string: ")
vowels = "aeiouAEIOU"
vowel_count = 0
for char in string:
  if char in vowels:
    vowel_count += 1
print(f"Number of vowels: {vowel_count}")
```
