### **1. squares numers:**
- Create a dictionary where keys are the first 5 natural numbers and values are their squares.
- Write a script to:
  - Print all keys and their respective values.
  - Add a new key-value pair: `6: 36`.
  - Check if a specific key (e.g., 4) exists in the dictionary.

---


```python
squares = {
	1: 1,
	2: 4,
	3: 9,
	4: 16,
	5: 25
}

for key, value in squares.items():
	print(key, ":", value)

squares[6] = 36

if 4 in squares:
	print("Key 4 exists")
else:
	print("Key 4 does not exist")
```

### **2. Student Grades:**
- Create a dictionary to store student names as keys and a list of their grades as values.
- Write a script to:
  - Calculate the average grade for each student.
  - Find the student with the highest average.


```python
students = {
	"Ali": [80, 90, 85],
	"Sara": [95, 92, 90],
	"Omar": [70, 75, 72]
}

highest_avg = 0

for name, grades in students.items():
	avg = sum(grades) / len(grades)
	print(f"{name} average = {avg:.2f}")

	if avg > highest_avg:
		highest_avg = avg

print(f"Highest average: {highest_avg:.2f}")
```


### **3. Shopping Cart:**
- Implement a shopping cart using a dictionary where keys are product names and values are their prices.
- Write a script to:
  - Add items to the cart.
  - Calculate the total price of items in the cart.

---


```python
cart = {}

cart["laptop"] = 1500
cart["pc"] = 1000
cart["tablet"] = 500

total = 0

for price in cart.values():
  total += price

print("Total price:", total)
```


### **4. Employee Database:**
- Create a list of dictionaries where each dictionary represents an employee with attributes:
  - `name`, `age`, and `department`.
- Write a script to:
  - Filter employees by a specific department.
  - Find the average age of employees.


```python
employees = [
	{"name": "Abdulrahman", "age": 24, "department": "CSE"},
	{"name": "Ahmed", "age": 25, "department": "IT"},
	{"name": "Omar", "age": 35, "department": "IT"},
	{"name": "Mohamed", "age": 28, "department": "OS"}
]

dept = input("Enter department: ")

print("Employees in", dept)
for emp in employees:
	if emp["department"] == dept:
		print(emp["name"])

total_age = 0

for emp in employees:
	total_age += emp["age"]

avg_age = total_age / len(employees)

print("Average age of employees:", f"{avg_age:.2f}")
```
